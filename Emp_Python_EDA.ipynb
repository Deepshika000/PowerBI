# **Explotatory Data Analysis**

```python
#Identifying Columns Name
dt.columns.tolist()
```

```python
#Employee Count by department
dt['Department'].value_counts()
```

```python
#Name of the Department with No.of Employees
c1=dt['Department'].value_counts()
print("Highest No.of Employee- ",c1.max())
print("Department with Highest No.of Employees is",c1.idxmax())
print()
c1=dt['Department'].value_counts()
print("Lowest No.of Employee- ",c1.min())
print("Department with Lowest No.of Employees is",c1.idxmin())
```

```python
#Employee counts by Gender
dt['Gender'].value_counts()
```

```python
#Average Age of Employee
print("Average Age of the Empoyee ",dt['Age'].mean().round())
```

```python
#Calculating Highest & Lowest Salary of the Employee
print(" Employee's Highest Salary- ",dt['Monthly_Salary'].max())
print(" Employee's Lowest Salary- ",dt['Monthly_Salary'].min())
print(" Average Salary of the Employees- ",dt['Monthly_Salary'].mean().round())
```

```python
#Average Performance Score
print("Average Performance Score of the Employee's ",dt['Employee_Satisfaction_Score'].mean().round())
```

```python
dt.groupby('Department')['Monthly_Salary'].mean().round()
```

```python
#Employee Based on their Education Level
print("Count of Employee Based on their Education Level",dt.groupby('Education_Level')['Employee_ID'].count())
```

```python
print("Average Performance Score By Department Wise",dt.groupby('Department')['Performance_Score'].mean().round())
```
